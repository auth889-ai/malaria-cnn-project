# preprocessing function:

```python
def resize_rescale(sample):

    image = sample["image"]
    label = sample["label"]

    image = tf.image.resize(
        image,
        (IM_SIZE, IM_SIZE)
    )

    image = image / 255.0

    return image, label
```

```python
train_processed = train_dataset.map(
    resize_rescale
)
```
